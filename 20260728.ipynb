```python
#생성자 작성하기 연습문제

class Product:
# 생성자 작성
    def __init__(self,name,code,count):
        self.name = name
        self.code = code
        self.count = count

    def show_info(self):
        return 'name=%s, code=%s, count=%s' % (
            self.name, self.code, self.count)

p1 = Product('clock', 'c0001', 500)
print(p1.show_info())

p2 = Product('tv', 't0001', 1500)
print(p2.show_info())
```

```python
#Letter클래스 작성하기 연습문제

class Letter:
# 생성자 작성, 변수 초기화
    def __init__(self):
        self.data = ''

# write 메소드 작성
    def write(self, data):
        self.data += data
        
# show 메소드 작성
    def show(self):
        return self.data

letter1 = Letter()
letter1.write('가나다라')
letter1.write('카타파하')
print(letter1.show())
letter2 = Letter()
letter2.write('ABCD')
letter2.write('WXYZ')
print(letter2.show())
```

```python
# Car 클래스 작성

class Car:
# 생성자 작성, 변수 초기화
    def __init__(self):
        self.opt = []
# add_option 메소드 작성
    def add_option(self, opt):
        self.opt.append(opt)

# show_option 메소드 작성
    def  show_option(self):
        return self.opt
        


car1 = Car()  # 클래스를 메모리에 객체로 생성
car2 = Car()  # Car의 인스턴스



car1.add_option('전동 트렁크')
car1.add_option('통풍 시트')
print(car1.show_option())
car2.add_option('뒷자리 에어백')
car2.add_option('하이패스')
print(car2.show_option())
```

```python
#Window를 상속받은 후 커스터마이징 예시

import sys
from PyQt5.QtWidgets import *
class MyWindow(QMainWindow):
    def __init__(self):
        super().__init__()
        self.btn = QPushButton("종료", self)

class MyWindow(QMainWindow):
    def __init__(self):
        super().__init__()
        self.btn = QPushButton("종료", self)
        self.btn.move(20, 20)
        self.btn.clicked.connect(self.btn_clicked)
    def btn_clicked(self):
        self.close()


app = QApplication.instance()
if not app:
 app = QApplication(sys.argv)

mywindow = MyWindow()
mywindow.show()

app.exec_()
```

```python
# 쇼핑몰 정보 및 상품 현황 출력하기 연습문제
import requests
res= requests.get('http://ggoreb.com/python/api/request_market.jsp')
parse = res.json()


print(f'[{parse["market_name"]}상품 현황 - 총 {parse["total_count"]}개 ]')
for i in parse["items"]:
    print(f'-상품명: {i["title"]} | 가격: {i["price"]}') 
```

```python
#전국 실시간 미세먼지 및 대기질 현황 출력하기 연습문제

import requests
res= requests.get('http://ggoreb.com/python/api/request_weather.jsp')
parse = res.json()


print(f'[{parse["data_source"]} - {parse["update_time"]} 기준 위험 지역 안내]')
for i in parse["regional_air_quality"]:
    if i["status"]["state"] == "나쁨":
        print(f' 경보 지역: {i["location"]} | 미세먼지 농도: {i["pm10"]} (상태: {i["status"]["state"]})') 
```

```python
# 실시간 영화 예매 차트 현황 및 평점 계산/출력하기
import requests
res= requests.get('http://ggoreb.com/python/api/request_movie.jsp')
parse = res.json()

print(f'[{parse["theater"]} 현재 순위]')
for i in parse["weekly_rank"]:
    print(f'[{i['rank']}위] {i["title"]} | 예매율: {i["booking_rate"]}% (평점:{i["score"]["star"]} 점)')

movie_list = parse["weekly_rank"]
total_score = sum([i["score"]["star"] for i in movie_list])
avg_score = total_score / len(movie_list)

print(f'현재 상영작 평균 평점: {avg_score:.2f}점')
```
