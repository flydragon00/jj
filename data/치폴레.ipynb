### Step 5. item_price에서 $를 없애고 float 형식으로 새롭게 저장해주세요.

```python
def price (x):
    return float(x[1:])

chipo['item_price'] = chipo['item_price'].apply(price)
chipo
```

### Step 6. 아이템별 주문 수를 출력해주세요

### Step 7. 아이템별 주문 총량을 출력해주세요.

### Step 8. 각 아이템의 가격을 구해서 unit_price 컬럼에 추가해주세요.
주문당 quantity가 1일때의 가격 (item_price / quantity)

```python
chipo['unit_price'] = chipo['item_price']/ chipo['quantity']
chipo
```

### Step 9. 아이템별 개당 가격의 평균을 구하고 가장 비싼순으로 5개를 출력해주세요

```python
result = chipo.groupby('item_name')['unit_price'].mean()
result.sort_values(ascending = False).head()
```

### Step 10. 한 주문(같은 order_id)에 20달러 이상 지불한 손님을 출력해주세요.

```python
chipo_price = chipo.groupby('order_id')['item_price'].sum()
chipo_price_index = chipo_price[ chipo_price >= 20 ].index
```

### Step 11. 위 10번에서 주문량이 2개인 아이템중 가장 많이 팔린 아이템은 무엇인가요?
Chips and Guacamole

```python
# chipo_price_index > 총 주문 금액이 20달러 이상인 주문 번호
chipo2 = chipo[chipo['order_id'].isin(chipo_price_index)]
```

### 13. Chicken이 포함된 음식은 몇 가지가 있나요?

```python
chipo[chipo['item_name'].str.contains('Chicken')]['item_name'].nunique()
```

```python
import re
def check_Chicken(x):
    return bool(re.search('Chicken',x))

chipo[chipo['item_name'].apply(check_Chicken)]['item_name'].unique()
```
