```python
# 정규 표현식 사용을 위한 모듈 import
import re
# 정규 표현식의 규칙 지정
pattern = re.compile(r'^[0-9]') # []가없으면 정확하게 그단어 , 있으면 그 단어중 하나
# 정규 표현식 규칙에 문자열을 입력하여 확인
print(pattern.search('안녕abc123'))
```

```python
#정규 표현식 연습문제
import re

pattern1 = re.compile(r'^[a-z A-Z]')
pattern2 = re.compile(r'[가-힣]')

# "안녕abc12" <= 영어0 숫자0

user_id = '12abc'

result1 = pattern1.search(user_id)
result2 = pattern2.search(user_id)
print(result1)
if result1 and (bool(result2)==False):
    print(f'{user_id} <- 가입 가능')
else:
    print(f'{user_id} <- 가입 불가')
```

```python
import re
text = "apple, ape, apricot, banana, apppple"
print(re.search(r"a.p", text))
# 'a' + 임의의 문자 + 'p'
# → <re.Match object; span=(0, 3), match='app'>
print(re.search(r"a[bc]e", "ace abe ace"))
# a 다음에 b나 c가 오고 e로 끝
# → <re.Match object; span=(0, 3), match='ace'>
print(re.search(r"ap*le", text))
# 'a' + 'p' 0회 이상 + 'le'
# → <re.Match object; span=(0, 5), match='apple'>
print(re.search(r"ap+le", text))
# 'a' + 'p' 1회 이상 + 'le'
# → <re.Match object; span=(0, 5), match='apple'>
print(re.search(r"ap{2}le", text))
# 'a' + 'p' 2번 + 'le'
# → <re.Match object; span=(0, 5), match='apple'>
print(re.search(r"ap?le", text))
# 'p'가 0번 또는 1번
# → None
```

```python
#주민등록번호 유효성 검사 연습문제
import re
pattern = re.compile(r'[0-9]{6}-[1-4][0-9]{6}')
text = '791111-1234567'

print(text, end=' <- ')

result = pattern.search(text)

if result:
    print('정상')
else:
    print('올바르지 못한 주민등록번호')
```

```python
#정규표현식 연습문제

import re
pattern = re.compile(r'^안녕하세요(.*)열심히(.*)감사합니다[.]', re.DOTALL)
text = '''안녕하세요. 저는 OOO에서 태어났고...
열심히 하겠습니다. 감사합니다.
'''
# 코드 작성 ( ② )
r= pattern.search(text)
print(pattern.search(text))
print(r.group(1))
print(r.group(2))
```

```python
#이름과 전화번호 찾기 연습문제
import re
pattern = re.compile(r'이름\s*:\s*(\w+),\s*전화번호\s*:\s*(\d+)')
text = '''
사용자 정보, 이름: 꼬렙, 전화번호 : 1234, 이메일 : seorab@naver.com
'''
# 코드 작성 ( ② )
r= pattern.search(text)
print(r.group(1))
print(r.group(2))
```

```python
#이메일 찾아내기 연습문제
import re
pattern = re.compile(r'이메일\s*:\s*(\w+@\w+[.]\w+)')
text = '''
사용자 정보, 이름: 꼬렙, 전화번호 : 1234, 이메일 : seorab@naver.com
'''
# 코드 작성 ( ② )
r = pattern.search(text)
print(r.group(0))
```

```python
#긍정 전방 탐색

import re

text = "user@example.com"

result = re.search(r"\w+(?=@)", text)
print(result) # <re.Match object; span=(0, 4), match='user'>

result2 = re.search(r"\w+[^@]", text)
print(result2) # <re.Match object; span=(0, 4), match='user'>

result3 = re.search(r"(\w+)@", text)
print(result3) # <re.Match object; span=(0, 5), match='user@'>

print(result3.group(1)) # user
```

```python
#ERROR 뒤의 숫자 코드만 확인
import re

text = """
INFO: 200 OK
ERROR: 404 Not Found
DEBUG: process started
ERROR: 500 Internal Server Error
"""

# 그룹 캡처 방식 → ERROR: 까지 패턴에 포함됨
pattern1 = re.compile(r'ERROR: (\d+)')
print("그룹 캡처:", pattern1.search(text))

# Lookbehind 방식 → ERROR: 는 결과에 포함되지 않음
pattern2 = re.compile(r'(?<=ERROR: )\d+')
print("탐색 사용:", pattern2.search(text))
```
