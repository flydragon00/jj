```python
#연습문제1)

url = "http://ggoreb.com/python/html/tiobe-index.html"
res = requests.get(url)
res.encoding = 'utf-8'
soup = bs(res.text, "html.parser")
trs= soup.select('#tiobe-tbody tr')


for tr in trs:
    rank = tr.select('.rank')[0].text
    lang = tr.select('.lang-name')[0].text
    rating = tr.select('.rating-badge')[0].text

    print(f'[{rank}위] {lang} | 점유율: {rating}')
```

```python
#연습문제2)
url = "http://ggoreb.com/python/html/dangn.html"
res = requests.get(url)
res.encoding = 'utf-8'
soup = bs(res.text, "html.parser")

cards = soup.select('#product-grid > .card')

for card in cards:
    title = card.select_one('.title-line').text
    price = card.select_one('.price').text

    print(f'{title} | {price}')
```

```python
#연습문제3)
url = "http://ggoreb.com/python/html/dangn.html"
res = requests.get(url)
res.encoding = 'utf-8'
soup = bs(res.text, "html.parser")
imgs = soup.select('div > a > div > img')
for img in imgs:

    p = 'http://ggoreb.com/python/html/' + img.get('src')
    r = requests.get(p)
    with open(img.get('src')[7:],'wb',) as f:
        f.write(r.content)
```
